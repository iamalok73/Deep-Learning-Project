```python
import numpy as np
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
import random
from skimage.util import random_noise
from skimage import transform
from cv2 import resize
```

```python
# pip install opencv-python
```

```python
img = mpimg.imread(r"C:\Users\AlokVerma\Desktop\Dog.jpg")
```

```python
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
new_img = resize(img,(150,150))
```

```python
plt.imshow(new_img)
plt.axis('off')
plt.show()
```

```python
## flipping the image
```

```python
hori = np.fliplr(img)
plt.imshow(hori)
plt.axis('off')
plt.show()
```

```python
ver = np.flipud(img)
plt.imshow(ver)
plt.axis('off')
plt.show()
```

```python
## Rotate the image.
```

```python
from skimage import transform
```

```python
new_rotate = transform.rotate(img,30)
plt.imshow(new_rotate)
plt.axis('off')
plt.show()
```

```python
new_rotate = transform.rotate(img,random.uniform(-60,40))
plt.imshow(new_rotate)
plt.axis('off')
plt.show()
```

```python
## Adding Noise 
```

```python
add_img = random_noise(img,mode='localvar',clip=True)
plt.imshow(add_img)
```
