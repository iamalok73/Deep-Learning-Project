```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
```

```python
df = pd.read_csv(r"C:\Users\AlokVerma\Desktop\ML DATA\spam.csv")
df.head(4)
```

```python
df["Spam"] = df["Category"].apply(lambda x : 1 if x =="spam" else 0 )
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df["Message"], df["Spam"], test_size=0.33, random_state=42)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
X_train.shape
```

```python
cv = CountVectorizer()
```

```python
X_train_cv = cv.fit_transform(X_train.values)
X_text_cv = cv.transform(X_test.values)
```

```python
X_train_cv.shape
```

```python
X_train_cv
```

```python
cv.get_feature_names_out()[5008]
```

```python
cv.vocabulary_
```

```python
X_train_arr = X_train_cv.toarray()
```

```python
X_train_arr
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
mm = MultinomialNB()
```

```python
mm.fit(X_train_cv,y_train)
```

```python
y_pred = mm.predict(X_text_cv)
```

```python
y_pred
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred))
```
