```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
corpus = [
    "Batman eats burger and Joker eats burger too",
    "Google releasing new pixel phone today",
    "Amazon launching new alexa device soon",
    "Rahul likes dosa but also eats burger"
]
```

```python
tf = TfidfVectorizer()
```

```python
tf.fit(corpus)
```

```python
transforms = tf.transform(corpus)
```

```python
transforms
```

```python
print(tf.vocabulary_)
```

```python
all_feature = tf.get_feature_names_out()[5]
all_feature
```

```python
for i in all_feature:
    txt = tf.vocabulary_.get(i)
    idf_score = tf.idf_[txt]
    print(f"{i}  = {idf_score}")
```

```python
print(transforms.toarray())
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv(r"C:\Users\AlokVerma\Desktop\ML DATA\Ecommerce_data.csv")
df.head()
```

```python
df["Label_num"] = df["label"].map({
    "Household" : 1,
    "Electronics" : 2,
    "Clothing & Accessories" :3,
    "Books":4
}
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df["Text"],df["Label_num"], test_size=0.33, random_state=42)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
df.shape
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tf = TfidfVectorizer()
```

```python
X_train_tf = tf.fit_transform(X_train)
X_test_tf = tf.transform(X_test)
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(X_train_tf,y_train)
```

```python
y_pred = model.predict(X_test_tf)
```

```python
y_pred 
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
